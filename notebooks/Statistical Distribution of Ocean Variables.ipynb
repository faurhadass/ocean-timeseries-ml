```python
import matplotlib.pyplot as plt
import numpy as np

# 1. Define parameters
mean = 0  # Center of the distribution
std_dev = 1  # Spread of the distribution
num_samples = 1000  # Number of data points

# 2. Generate random normal distribution data
data = np.random.normal(mean, std_dev, num_samples)

# 3. Plot the histogram of the data
plt.hist(data, bins=30, density=True, alpha=0.6, color="g", edgecolor="black")

# 4. Plot the theoretical probability density function (PDF)
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = (1 / (std_dev * np.sqrt(2 * np.pi))) * np.exp(
    -0.5 * ((x - mean) / std_dev) ** 2
)
plt.plot(x, p, "k", linewidth=2)

# 5. Add labels and title
plt.title(
    f"Normal Distribution ($\\mu={mean}$, $\\sigma={std_dev}$)", fontsize=14
)
plt.xlabel("Value", fontsize=12)
plt.ylabel("Density", fontsize=12)
plt.grid(axis="y", alpha=0.75)

# 6. Show plot
plt.show()
```

# Testing out the Central Limit Theorem

```python
mean_val = np.mean(d["temp"])
print(mean_val);
median_val = np.median(d["temp"])
print(median_val);
# Sample Variance and Std Dev (ddof=1)
variance_val = np.var(d["temp"], ddof=1)
print(variance_val);
stdev_val = np.std(d["temp"], ddof=1)
print(stdev_val);


plt.hist(d["temp"].flatten(), bins=30, density=True, alpha=0.6, color= 'g', edgecolor = 'b')

# 4. Plot the theoretical probability density function (PDF)
xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = (1 / (stdev_val * np.sqrt(2 * np.pi))) * np.exp(
    -0.5 * ((x -mean_val) / stdev_val) ** 2
)

plt.title(
    f"Temp Speed Distribution ($\\mu={int(mean_val)}$, $\\sigma={int(stdev_val)}$)", fontsize=14
)
plt.plot(x, p, "k", linewidth=2)
plt.xlabel("Value", fontsize=12)
plt.ylabel("Density", fontsize=12)

plt.show()
```
